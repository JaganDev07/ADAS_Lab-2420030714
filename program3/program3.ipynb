# Program 3 — Forward Collision Warning System

**Aim:** Implement and evaluate a Forward Collision Warning System that alerts the
driver when the distance to the vehicle ahead becomes unsafe.

**Method:** The safe following distance is the distance the car travels during the
driver's reaction time. Each distance reading is compared against it, and the
time-to-collision (TTC) is reported.

```python
# Own vehicle state
speed_kmph = 60.0          # own vehicle speed
reaction_time = 1.5        # driver reaction time in seconds

speed_mps = speed_kmph * 1000 / 3600
safe_distance = speed_mps * reaction_time

print(f"Speed         : {speed_kmph} km/h ({speed_mps:.1f} m/s)")
print(f"Safe distance : {safe_distance:.1f} m")
```

```python
# Distance to the vehicle ahead (in metres), one reading per frame
distances = [80.0, 60.0, 45.0, 30.0, 20.0, 12.0, 6.0]
distances
```

```python
for distance in distances:

    time_to_collision = distance / speed_mps

    if distance < safe_distance * 0.5:
        warning = "CRITICAL: COLLISION WARNING!"
    elif distance < safe_distance:
        warning = "CAUTION: REDUCE SPEED"
    else:
        warning = "SAFE"

    print(f"Distance: {distance:5.1f} m   TTC: {time_to_collision:4.1f} s   ->  {warning}")
```
